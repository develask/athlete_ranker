```python
import json
import sys
import numpy as np

from pathlib import Path
repo_root = Path.cwd().parent if (Path.cwd().parent / "util").is_dir() else Path.cwd()
sys.path.insert(0, str(repo_root))

from util.build_dataset import build_athlete_records, athlete_profile
```

```python
data = json.load(open('../data/processed/dataset.json'))
athlete_records = build_athlete_records(data)
athlete_profiles = {athlete_id: athlete_profile(records) for athlete_id, records in athlete_records.items()}
```

```python
total = 0
min5 = 0
mixed = []
for athlete_id, profile in athlete_profiles.items():
    probs, count = profile['preferred_embarcacion']
    if count == 0:
        continue
    counts = [probs[0] * count, probs[1] * count, probs[2] * count]
    if len([c for c in counts if c > 5]) >= 2:
        min5 += 1
        mixed.append((probs, count))
    total += 1

min5, total
```

```python
sorted(mixed, key=lambda x: max(x[0]), reverse=True)
```
