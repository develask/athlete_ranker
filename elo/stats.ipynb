```python
import json
from collections import Counter
from pathlib import Path
```

```python
def distribution(title, values, limit=None):
    counts = Counter("(blank)" if value in (None, "") else value for value in values)

    def sort_key(item):
        value = item[0]
        if value == "(blank)":
            return (1, "")
        if isinstance(value, (int, float)):
            return (0, value)
        return (0, str(value).casefold())

    print(f"\n{title} (n={sum(counts.values())})")
    items = sorted(counts.items(), key=sort_key)
    if limit:
        items = items[:limit]
    for value, count in items:
        print(f"  {value}: {count} ({count / sum(counts.values()):.1%})")
    if limit and len(counts) > limit:
        print(f"  ... {len(counts) - limit} more values")
```

```python
path = Path.cwd().parent / "data/processed/dataset.json"
with path.open(encoding="utf-8") as file:
    regatas = json.load(file)

pruebas = [prueba for regata in regatas for prueba in regata.get("pruebas", [])]
results = [result for prueba in pruebas for result in prueba.get("results", [])]

print(f"File: {path}")
print(f"Regatas: {len(regatas)}")
print(f"Pruebas: {len(pruebas)}")
print(f"Results: {len(results)}")

print("\n=== PRUEBAS ===")
for field in (
    "tipo", "distancia_exacta", "sexo", "categoria",
    "embarcacion_tipo", "embarcacion_num",
):
    distribution(field, (prueba.get(field) for prueba in pruebas))
```

```python
print("\n=== REGATAS ===")
distribution("Year", (str(regata.get("fecha", ""))[:4] for regata in regatas))
distribution("League", (regata.get("regata_liga") for regata in regatas))
distribution("Organizer", (regata.get("organizador_name") for regata in regatas), limit=20)
distribution("Pruebas per regata", (len(regata.get("pruebas", [])) for regata in regatas))
"""
distribution(
    "Results per regata",
    (sum(len(prueba.get("results", [])) for prueba in regata.get("pruebas", [])) for regata in regatas),
)"""
```
